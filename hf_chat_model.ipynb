```python
# prompt: use the pipe for chat

from llama_cpp import Llama
from transformers import pipeline
!pip install -q llama-cpp-python

llm = Llama.from_pretrained(
	repo_id="mradermacher/Flirty-Mistral-GGUF",
	filename="Flirty-Mistral.IQ3_M.gguf",
)

llm.create_chat_completion(
    messages = [{"role": "user", "content": "Hi"}]
)

llm = Llama.from_pretrained(
    repo_id="mradermacher/Flirty-Mistral-GGUF",
    filename="Flirty-Mistral.Q2_K.gguf",  # Use a more efficient quantized model
    n_threads=8,                            # Adjust based on CPU
    n_gpu_layers=20,                        # Allocate layers to GPU (if available)
    n_ctx=1024,                             # Reduce context size
    use_mmap=True,
    use_mlock=True
)

response = llm.create_chat_completion(
    messages=[{"role": "user", "content": "Hi"}]
)

print(response)
# Use a pipeline as a high-level helper

pipe = pipeline("text-generation", model="mylesfriedman30/llama-2-7b-flirtbot")

# Example using the pipeline with user input
user_input = "How are you doing today?"
result = pipe(user_input)
print(result[0]['generated_text'])


# Example chat loop using the pipeline
while True:
  user_input = input("User: ")
  if user_input.lower() in ["quit", "exit", "bye"]:
    break
  result = pipe(user_input)
  print("Bot:", result[0]['generated_text'])
```
